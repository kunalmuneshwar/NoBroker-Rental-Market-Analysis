```python
import pandas as pd
import numpy as np

def clean_price(price):
    if pd.isna(price):
        return np.nan

    price = str(price).strip()

    if price == "Call for Price":
        return np.nan

    if "Lac" in price:
        num = float(price.replace("Lac", "").strip())
        return num * 100000

    price = price.replace(",", "")

    try:
        return float(price)
    except:
        return np.nan
```

```python
df['Price'] = df['Price'].apply(clean_price)
```

```python
import pandas as pd
import numpy as np

def clean_area(area):
    if pd.isna(area):
        return np.nan

    area = str(area).strip()

    if area.lower() == 'none':
        return np.nan

    try:
        return float(area.split()[0])
    except:
        return np.nan
```

```python
df['Area'] = df['Area'].apply(clean_area)
```

```python
df['Area'].isnull().sum()
```

```python
df['Price_per_sqft'] = df['Price'] / df['Area']
```
